```python
import random

def objective_function(x):
    """Example objective function to maximize: -(x-10)^2 + 100"""
    return -(x - 10)**2 + 100

def generate_neighbors(current_solution, step_size=0.1):
    """Generates neighboring solutions by slightly perturbing the current solution."""
    neighbors = []
    neighbors.append(current_solution + step_size)
    neighbors.append(current_solution - step_size)
    return neighbors

def hill_climbing(objective_function, generate_neighbors, initial_solution, max_iterations=1000):
    """Implements the Hill Climbing algorithm.

    Args:
        objective_function: A function to maximize.
        generate_neighbors: A function that generates neighbors for a given solution.
        initial_solution: The starting point for the search.
        max_iterations: Maximum number of iterations to prevent infinite loops.

    Returns:
        The best solution found (local optimum).
    """
    current_solution = initial_solution
    current_value = objective_function(current_solution)

    print(f"Initial solution: {current_solution:.2f}, Value: {current_value:.2f}")

    for i in range(max_iterations):
        neighbors = generate_neighbors(current_solution)
        best_neighbor = current_solution
        best_neighbor_value = current_value

        for neighbor in neighbors:
            neighbor_value = objective_function(neighbor)
            if neighbor_value > best_neighbor_value:
                best_neighbor_value = neighbor_value
                best_neighbor = neighbor

        # If no better neighbor is found, we've reached a local optimum
        if best_neighbor == current_solution:
            print(f"Reached local optimum at iteration {i+1}.")
            break
        else:
            current_solution = best_neighbor
            current_value = best_neighbor_value
            print(f"Iteration {i+1}: New solution: {current_solution:.2f}, Value: {current_value:.2f}")

    print("\nHill Climbing finished.")
    print(f"Local optimum found at x = {current_solution:.2f} with value = {current_value:.2f}")
    return current_solution, current_value

# --- Example Usage ---
if __name__ == "__main__":
    # Set a random initial solution within a reasonable range
    initial_x = random.uniform(0, 10)

    print("Starting Hill Climbing...")
    best_x, best_value = hill_climbing(objective_function, generate_neighbors, initial_x)
```
