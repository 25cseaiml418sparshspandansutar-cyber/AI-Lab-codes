```python
def alpha_beta(node, depth, alpha, beta, maximizing_player):
  #Leaf Node
  if depth == 0 or isinstance(node, int):
    print(f"Leaf: {node}")
    return node

  if maximizing_player:
    best = float('-inf')
    print(f"\nMAX node | α = {alpha}, β = {beta}")

    for child in node:
      value = alpha_beta(child, depth-1, alpha, beta, False)

      best = max(best, value)
      alpha = max(alpha, best)

      print(f"MAX: value ={value}, α = {alpha}, β = {beta}")

      if beta <= alpha:
         print(">>> PRUNING remaining MAX children")
         break

    return best
  else:
    best = float('inf')
    print(f"\nMIN node | α = {alpha}, β = {beta}")

    for child in node:
      value = alpha_beta(child, depth-1, alpha, beta, True)

      best = min(best, value)
      best = min(best,value)

      print(f"MIN: value ={value}, α = {alpha}, β = {beta}")

      if beta <= alpha:
        print(">>> PRUNING remaining MIN children")
        break

    return best


#Example Tree
tree = [
    [3,5,2],
    [1,9,4]
]

#Start Alpha-Beta Pruning
result = alpha_beta(
    tree,
    depth=2,
    alpha=float('-inf'),
    beta=float('inf'),
    maximizing_player=True
)

print(f"\nFinal Answer: ", result)
```
