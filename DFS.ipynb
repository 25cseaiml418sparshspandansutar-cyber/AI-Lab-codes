```python

# DFS function
def dfs(graph, node, visited):
    if node not in visited:
        print(node, end=" ")
        visited.add(node)

        # Visit all adjacent nodes
        for neighbor in graph[node]:
            dfs(graph, neighbor, visited)

# Example graph (Adjacency List)
graph = {}

n = int(input("enter the number of vertices: "))

for i in range(n):
  vertex = input("Enter the value {i+1}: ")
  neighbors = input("Enter adjacent vertices of {vertex}(space-separated): ").split()
  graph[vertex] = neighbors


  start = input("Enter the starting vertex: ")

  print("\nGraph:", graph)
  print("DFS Travesal: ", end=" ")

visited = set()

print("DFS Traversal:")
dfs(graph, 'A', visited)
```
