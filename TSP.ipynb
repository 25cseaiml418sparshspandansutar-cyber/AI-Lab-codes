```python
from itertools import permutations

def tsp(graph, start):
    n = len(graph)
    vertices = [i for i in range(n) if i != start]

    min_cost = float('inf')
    best_path = []

    for path in permutations(vertices):
        cost = 0
        k = start

        for vertex in path:
            cost += graph[k][vertex]
            k = vertex

        cost += graph[k][start]

        if cost < min_cost:
            min_cost = cost
            best_path = [start] + list(path) + [start]

    return min_cost, best_path


# Distance matrix
graph = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

cost, path = tsp(graph, 0)

print("Minimum Cost:", cost)
print("Optimal Path:", path)
```
