```python
import numpy as np

x = np.array([1, 2, 3])
y = np.array([2, 4, 5])
lr = 0.01

w = 0
b = 0

#ITERATION 1
y_pred = w * x + b

cost = np.mean((y_pred - y) ** 2) / 2

dw = np.mean((y_pred - y) * x)
db = np.mean(y_pred - y)

w = w - lr * dw
b = b - lr * db

print("Iteration 1")
print("w =", w)
print("b =", b)
print("cost =", cost)


#ITERATION 2 `
y_pred = w * x + b

cost = np.mean((y_pred - y) ** 2) / 2

dw = np.mean((y_pred - y) * x)
db = np.mean(y_pred - y)

w = w - lr * dw
b = b - lr * db

print("\nIteration 2")
print("w =", w)
print("b =", b)
print("cost =", cost)

print("\nFinal prediction:", w * x + b)
```
