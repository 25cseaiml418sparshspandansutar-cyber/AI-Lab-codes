```python
N = 8
board = [-1] * N

def is_safe(row, col):
    i = 0

    while i < row:
        if board[i] == col:
            return False

        if abs(board[i] - col) == abs(i - row):
            return False

        i += 1

    return True


def solve():
    row = 0

    while row < N:
        col = board[row] + 1

        while col < N and not is_safe(row, col):
            col += 1

        if col < N:
            board[row] = col
            row += 1
        else:
            board[row] = -1
            row -= 1

            if row < 0:
                break
#MAIN PROGRAM.

def print_board():
    row = 0

    while row < N:
        col = 0

        while col < N:
            if board[row] == col:
                print("Q", end=" ")
            else:
                print(".", end=" ")

            col += 1

        print()
        row += 1


solve()
print_board()
```
