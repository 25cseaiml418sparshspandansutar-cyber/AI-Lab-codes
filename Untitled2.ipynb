```python
import pandas as pd
import matplotlib.pyplot as plt

# --------------------------------------------------
# 1. Set the plotting style
# --------------------------------------------------
plt.style.use('seaborn-v0_8-whitegrid')

# --------------------------------------------------
# 2. Load the dataset
# --------------------------------------------------
file_path = r'D:\pythonprogs\student-mat.csv'

try:
    df = pd.read_csv(file_path)
except FileNotFoundError:
    print("Error: Dataset file was not found.")
    print("Please check the file path:")
    print(file_path)
    exit()

# --------------------------------------------------
# 3. Display basic information
# --------------------------------------------------

print("=" * 50)
print("FIRST 5 ROWS")
print("=" * 50)
print(df.head())

print("\n" + "=" * 50)
print("DATASET SHAPE")
print("=" * 50)
print(df.shape)

print("\n" + "=" * 50)
print("COLUMN NAMES")
print("=" * 50)
print(df.columns.tolist())

print("\n" + "=" * 50)
print("DATASET INFORMATION")
print("=" * 50)
df.info()

print("\n" + "=" * 50)
print("MISSING VALUES")
print("=" * 50)
print(df.isnull().sum())

print("\n" + "=" * 50)
print("BASIC STATISTICS")
print("=" * 50)
print(df.describe())

# --------------------------------------------------
# 4. Histogram of final grades
# --------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    df['G3'],
    bins=10,
    color='steelblue',
    edgecolor='black'
)

plt.title('Distribution of Final Grades')
plt.xlabel('Final Grade (G3)')
plt.ylabel('Number of Students')

plt.tight_layout()
plt.show()

# --------------------------------------------------
# 5. Study time vs final grade
# --------------------------------------------------

plt.figure(figsize=(8, 5))

plt.scatter(
    df['studytime'],
    df['G3'],
    color='darkorange',
    alpha=0.7,
    edgecolors='black'
)

plt.title('Study Time vs Final Grade')
plt.xlabel('Study Time')
plt.ylabel('Final Grade (G3)')

plt.tight_layout()
plt.show()

# --------------------------------------------------
# 6. Absences vs final grade
# --------------------------------------------------

plt.figure(figsize=(8, 5))

plt.scatter(
    df['absences'],
    df['G3'],
    color='seagreen',
    alpha=0.7,
    edgecolors='black'
)

plt.title('Absences vs Final Grade')
plt.xlabel('Number of Absences')
plt.ylabel('Final Grade (G3)')

plt.tight_layout()
plt.show()

# --------------------------------------------------
# 7. Average grade by gender
# --------------------------------------------------

gender_avg = df.groupby('sex')['G3'].mean()

plt.figure(figsize=(7, 5))

gender_avg.plot(
    kind='bar',
    color=['cornflowerblue', 'lightcoral'],
    edgecolor='black'
)

plt.title('Average Final Grade by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Final Grade')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()

# --------------------------------------------------
# 8. Correlation matrix
# --------------------------------------------------

numeric_df = df.select_dtypes(include='number')

correlation = numeric_df.corr()

plt.figure(figsize=(12, 9))

plt.imshow(
    correlation,
    cmap='coolwarm',
    interpolation='none',
    aspect='auto'
)

plt.colorbar(label='Correlation')

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title('Correlation Matrix')

plt.tight_layout()
plt.show()

# --------------------------------------------------
# 9. Correlation with final grade (G3)
# --------------------------------------------------

print("\n" + "=" * 50)
print("CORRELATION WITH FINAL GRADE (G3)")
print("=" * 50)

g3_correlation = (
    correlation['G3']
    .sort_values(ascending=False)
)

print(g3_correlation)

# --------------------------------------------------
# 10. Program completed
# --------------------------------------------------

print("\nProgram completed successfully!")
```
