```python
from collections import deque

#Function to solve Water Jug problem using BFS
def water_jug(capacity1, capacity2, target):
  visited = set()
  queue = deque()

  #Initial state: both jugs are empty
  queue.append((0,0,[]))

  while queue:
    jug1, jug2, path = queue.popleft()
    #skip if the state has already been visited
    if(jug1, jug2) in visited:
      continue

    visited.add((jug1, jug2))

    #Add current state to the path
    current_path = path + [(jug1, jug2)]

    #Check if target is reached
    if jug1 == target or jug2 == target:
      return current_path

    #Generate all possible next states
    next_states = [
        # Fill Jug 1
        (capacity1, jug2),

        #Fill Jug 2
        (jug1, capacity2),

        # EMPTY Jug1
        (0, jug2),

        # Empty Jug 2
        (jug1, 0),

        #Pour Jug1 --> Jug2
        (
            jug1 - min(jug1, capacity2 - jug2),
            jug2 + min(jug1, capacity2 - jug2)
        ),

        #Pour Jug2 --> Jug1
        (
            jug1 + min(jug2, capacity1 - jug1),
            jug2 - min(jug2, capacity1 - jug1)
        )
    ]

    #Add unvisited states to the queue
    for state in next_states:
      if state not in visited:
        queue.append((state[0], state[1], current_path))

  #No solution found
  return None

#Main Program
jug1= int(input("Enter capacity of Jug 1: "))
jug2= int(input("Enter capacity of Jug 2: "))
target= int(input("Enter target capacity: "))

solution = water_jug(jug1, jug2, target)

if solution:
  print("\nSteps to reach the target capacity: ")

  for step in solution:
    print(step)
else:
  print("No solution exists.")
```

Water jug Problem

Aim

To implement the water jug problem using BFS in python ad find a sequence of operations to obtain the desired target quantity of water.

Theory

The Water Jug Problem is a classical problem in Artificial Intelligence that demonstrates state-space search. It involves two jugs with known capacities and no measurement markings. The objective is to obtain a specific amount of water using a set of allowed operations.
